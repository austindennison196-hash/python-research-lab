# Structural complexity and constrained emergence

**DEFINITION.** For a uniform distribution over $N$ distinct states, the state entropy is $H=\log_2N$ bits. This differs from the symbol entropy of one binary string. **HYPOTHESIS.** Stabilization might be modeled as restrictions on allowed transitions; here a bounded random walk is only a toy operationalization.

Both unconstrained and constrained walks below share a seed. Neither simulation is a proposed law of physics.

```python
from pathlib import Path
import math
import os
import random
import sys
import tempfile

ROOT = Path.cwd()
if not (ROOT / "src" / "research_lab").is_dir():
    ROOT = ROOT.parent
assert (ROOT / "src" / "research_lab").is_dir(), "Run from the repository or notebooks directory"
sys.path.insert(0, str(ROOT / "src"))
os.environ.setdefault("MPLCONFIGDIR", str(Path(tempfile.gettempdir()) / "research-lab-mpl"))
SEED = 2026
random.seed(SEED)
import matplotlib.pyplot as plt
from IPython import get_ipython
from IPython.display import display
get_ipython().run_line_magic("matplotlib", "inline")
```

```python
from research_lab.structural_logic import generate_stages
from research_lab.information_theory import shannon_entropy, empirical_entropy
from research_lab.emergence import constrained_walk
from research_lab.visualization import plot_comparison

stages = generate_stages(6)
for n, stage in enumerate(stages):
    state_entropy = shannon_entropy([1/len(stage)]*len(stage))
    assert math.isclose(state_entropy, math.log2(len(stage)), abs_tol=1e-12)
    print(n, len(stage), state_entropy, empirical_entropy(stage[0]))
free = constrained_walk(100, seed=SEED)
bounded = constrained_walk(100, seed=SEED, bound=3)
assert max(map(abs, bounded)) <= 3
records = [{"generation": i, "free": x, "bounded": y} for i, (x,y) in enumerate(zip(free,bounded))]
fig = plot_comparison(records, metrics=("free", "bounded"))
display(fig)
```

## Interpretation and limitations

A selected constraint changes the reachable trajectory; the observed effect follows from that constraint and is not evidence for the cosmological hypothesis. Entropy depends on a specified probability measure. Uniform state entropy here is a comparison convention, not the distribution induced by choosing rewrite positions uniformly.
