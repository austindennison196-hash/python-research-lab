# Information theory

**DEFINITION.** Shannon entropy is $H(p)=-\sum_i p_i\log_2p_i$. Mutual information is zero for independent variables, while Kullback–Leibler divergence is nonnegative but not symmetric.

This experiment compares theoretical Bernoulli entropy with a fixed-seed empirical sample. It separates symbol frequencies from a distribution over generated structures.

```python
from pathlib import Path
import math
import os
import random
import sys
import tempfile

ROOT = Path.cwd()
if not (ROOT / "src" / "research_lab").is_dir():
    ROOT = ROOT.parent
assert (ROOT / "src" / "research_lab").is_dir(), "Run from the repository or notebooks directory"
sys.path.insert(0, str(ROOT / "src"))
os.environ.setdefault("MPLCONFIGDIR", str(Path(tempfile.gettempdir()) / "research-lab-mpl"))
SEED = 2026
random.seed(SEED)
import matplotlib.pyplot as plt
from IPython import get_ipython
from IPython.display import display
get_ipython().run_line_magic("matplotlib", "inline")
```

```python
from research_lab.information_theory import shannon_entropy, empirical_entropy, mutual_information, kl_divergence, lz_complexity
from research_lab.visualization import plot_function

rng = random.Random(SEED)
sample = [int(rng.random() < .5) for _ in range(2000)]
assert shannon_entropy([.5,.5]) == 1.
assert abs(mutual_information([[.25,.25],[.25,.25]])) < 1e-12
assert abs(kl_divergence([.3,.7],[.3,.7])) < 1e-12
print({"empirical_entropy_bits": empirical_entropy(sample), "lz_complexity": lz_complexity("".join(map(str,sample)))})
fig = plot_function(lambda p: shannon_entropy([p,1-p]), 0, 1)
display(fig)
```

## Interpretation and limitations

The entropy maximum reflects the Bernoulli model and base-2 logarithms. An empirical sample has sampling error. Lempel–Ziv complexity depends on the chosen parsing definition, sequence length, and representation; it is not a universal measure of emergence or physical complexity.
