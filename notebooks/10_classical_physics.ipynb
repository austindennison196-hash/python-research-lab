# Classical harmonic oscillator

**ESTABLISHED MODEL.** A point mass in an ideal linear spring satisfies $\ddot x+\omega^2x=0$ and conserves $E=\tfrac12mv^2+\tfrac12m\omega^2x^2$. Here $x(0)=1$ m, $v(0)=0$ m/s, $m=1$ kg, and $\omega=2$ rad/s.

The exact solution is a reference for future numerical integrators, not a simulation of all real springs.

```python
from research_lab.physics.classical import harmonic_oscillator, oscillator_energy
from research_lab.visualization import plot_function

omega = 2.
states = [harmonic_oscillator(t/20, omega=omega) for t in range(201)]
energies = [oscillator_energy(x, v, mass=1., omega=omega) for x,v in states]
assert max(energies)-min(energies) < 1e-12
print({"initial_energy_J": energies[0], "maximum_energy_drift_J": max(energies)-min(energies)})
fig = plot_function(lambda t: harmonic_oscillator(t, omega=omega)[0], 0, 10)
display(fig)
```
