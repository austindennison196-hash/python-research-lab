# Structural logic and its definitions

The original interpretation uses $0$ for potential or space and $1$ for distinction. Its relationships are $0\leftarrow1$ and $1\rightarrow0$.

The literal correspondence is `0 ↔ x`, `10 ↔ (x)`, `1100 ↔ ((x)x)`, `111000 ↔ (((x)x)x)`, and `110100 ↔ ((x)(x)x)`.

**DEFINITION.** The implemented grammar extends these examples by writing a node as `(`, its ordered children, and `x)`. The seed is the special expression `x`. This extension is a candidate formalization, not a uniquely inferred meaning of the symbols.
