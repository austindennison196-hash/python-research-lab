# Symbolic mathematics

For $f(x)=x^3-3x$, symbolic differentiation gives $f'(x)=3x^2-3$, while integration and factorization expose different mathematical relationships. The reusable symbolic report computes these transformations through SymPy.

**METHOD.** Use explicit variable names and domain assumptions when interpreting symbolic expressions. This example uses a polynomial on the real line.

```python
from research_lab.mathematics import symbolic_report
from research_lab.visualization import plot_function

report = symbolic_report("x**3 - 3*x", variable="x")
for operation, value in report.items():
    print(operation, value)
import sympy as sp
x = sp.Symbol("x")
assert sp.simplify(report["derivative"] - (3*x*x - 3)) == 0
fig = plot_function(lambda t: t**3 - 3*t, -2, 2)
display(fig)
```

## Interpretation and limitations

Symbolic identities are exact within their assumptions; a numerical plot only illustrates selected real values. For transcendental expressions, branch cuts, singularities, and complex domains need explicit handling. Do not treat an unsimplified difference or failed solver as a disproof.
