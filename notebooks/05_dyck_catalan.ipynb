# Dyck words and Catalan numbers

**ESTABLISHED RESULT.** Standard Dyck words with $n$ pairs are counted by $C_n=\frac{1}{n+1}\binom{2n}{n}$. Candidate `zero_split` produces primitive Dyck words at generation $n\ge1$, counted by $C_{n-1}$.

The size shift matters: a primitive word has one outer pair whose contents are an arbitrary Dyck word. The documentation proves the relationship under the explicit rewrite definition; this finite enumeration is a computational cross-check.

```python
from pathlib import Path
import math
import os
import random
import sys
import tempfile

ROOT = Path.cwd()
if not (ROOT / "src" / "research_lab").is_dir():
    ROOT = ROOT.parent
assert (ROOT / "src" / "research_lab").is_dir(), "Run from the repository or notebooks directory"
sys.path.insert(0, str(ROOT / "src"))
os.environ.setdefault("MPLCONFIGDIR", str(Path(tempfile.gettempdir()) / "research-lab-mpl"))
SEED = 2026
random.seed(SEED)
import matplotlib.pyplot as plt
from IPython import get_ipython
from IPython.display import display
get_ipython().run_line_magic("matplotlib", "inline")
```

```python
from research_lab.structural_logic import dyck_words, generate_stages
from research_lab.visualization import plot_comparison

catalan = lambda n: math.comb(2*n, n) // (n+1)
stages = generate_stages(7, rule="zero_split")
records = []
for n in range(1, 8):
    dyck_count = len(dyck_words(n))
    assert dyck_count == catalan(n)
    assert len(stages[n]) == catalan(n-1)
    records.append({"generation": n, "standard_dyck": dyck_count,
                    "candidate_primitive": len(stages[n])})
print(records)
fig = plot_comparison(records, metrics=("standard_dyck", "candidate_primitive"))
display(fig)
```

## Interpretation and limitations

The two counts differ at the same generation; an unqualified “the proposed system is the Catalan numbers” would conceal the restriction and index shift. The conditional theorem applies to the formal model, not to every interpretation of the originator's theory. Finite agreement alone is not the proof.
