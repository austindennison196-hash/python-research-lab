# Reproducible experimental hypothesis testing

**HYPOTHESIS.** A seeded Bernoulli sample with parameter $p=0.5$ has empirical symbol entropy close to one bit when enough samples are collected. A single finite sample cannot prove the distributional assertion.

Run the registered experiment twice with the same parameters and seed, retain independent run identities, and compare numerical outputs. Temporary directories keep this introductory notebook from polluting the research record.

```python
from pathlib import Path
import math
import os
import random
import sys
import tempfile

ROOT = Path.cwd()
if not (ROOT / "src" / "research_lab").is_dir():
    ROOT = ROOT.parent
assert (ROOT / "src" / "research_lab").is_dir(), "Run from the repository or notebooks directory"
sys.path.insert(0, str(ROOT / "src"))
os.environ.setdefault("MPLCONFIGDIR", str(Path(tempfile.gettempdir()) / "research-lab-mpl"))
SEED = 2026
random.seed(SEED)
import matplotlib.pyplot as plt
from IPython import get_ipython
from IPython.display import display
get_ipython().run_line_magic("matplotlib", "inline")
```

```python
from research_lab.experiments import run_experiment
from research_lab.visualization import plot_function
from research_lab.information_theory import shannon_entropy

parameters = {"samples": 2000, "probability": .5}
with tempfile.TemporaryDirectory() as directory:
    first = run_experiment("entropy-sampling", parameters, seed=SEED, output_dir=directory)
    second = run_experiment("entropy-sampling", parameters, seed=SEED, output_dir=directory)
    assert first.path.exists() and second.path.exists()
    assert first.record["outputs"] == second.record["outputs"]
    print({"outputs": first.record["outputs"], "validation": first.record["validation"]})
    print("Same numerical outputs; different run identities and timestamps are expected.")
fig = plot_function(lambda p: shannon_entropy([p,1-p]), 0, 1)
display(fig)
```

## Interpretation and limitations

This check demonstrates deterministic seeded numerical outputs and retained run metadata. It is not a formal significance test and does not prove that every random realization lies close to its expectation. For scientific inference, define the null model, uncertainty, sample size, decision threshold, and comparison procedure before inspecting the outcome.
