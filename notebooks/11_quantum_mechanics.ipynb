# Quantum two-level dynamics

**ESTABLISHED MODEL.** For a two-level system initially in $|0\rangle$ with Hamiltonian $H=\hbar\Omega\sigma_x/2$, ideal unitary dynamics give $P_1(t)=\sin^2(\Omega t/2)$ and $P_0(t)+P_1(t)=1$.

We compare the reusable analytic model with the optional QuTiP solver when installed. Binary labels here name basis states and are not an identification of quantum mechanics with the proposed structural system.

```python
from pathlib import Path
import math
import os
import random
import sys
import tempfile

ROOT = Path.cwd()
if not (ROOT / "src" / "research_lab").is_dir():
    ROOT = ROOT.parent
assert (ROOT / "src" / "research_lab").is_dir(), "Run from the repository or notebooks directory"
sys.path.insert(0, str(ROOT / "src"))
os.environ.setdefault("MPLCONFIGDIR", str(Path(tempfile.gettempdir()) / "research-lab-mpl"))
SEED = 2026
random.seed(SEED)
import matplotlib.pyplot as plt
from IPython import get_ipython
from IPython.display import display
get_ipython().run_line_magic("matplotlib", "inline")
```

```python
from research_lab.physics.quantum import two_level_probabilities, qutip_rabi
from research_lab.visualization import plot_function

times = [i*math.pi/20 for i in range(41)]
expected = [two_level_probabilities(t)[1] for t in times]
assert all(abs(sum(two_level_probabilities(t))-1) < 1e-12 for t in times)
try:
    simulated = qutip_rabi(times)
except ImportError as exc:
    print("Optional QuTiP comparison not executed:", exc)
else:
    deviation = max(abs(a-b) for a,b in zip(simulated, expected))
    assert deviation < 1e-4
    print({"maximum_probability_difference": deviation})
fig = plot_function(lambda t: two_level_probabilities(t)[1], 0, 2*math.pi)
display(fig)
```

## Interpretation and limitations

Probability normalization is tested explicitly; the QuTiP comparison is identified as optional rather than counted as passed when unavailable. The isolated two-level Hamiltonian omits decoherence, noise, and interactions. These results do not justify a cosmological or philosophical interpretation of the structural notation.
