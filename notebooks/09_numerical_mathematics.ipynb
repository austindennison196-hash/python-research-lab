# Numerical mathematics

We compare numerical calculations with analytic references: $\int_0^\pi\sin x\,dx=2$, a two-variable linear system, minimizing $(x-2)^2$, and solving $y'=-y$, $y(0)=1$.

**METHOD.** Numerical accuracy is tested with explicit tolerances; plots are not accuracy tests. These small, deterministic examples require the scientific extra.

```python
from pathlib import Path
import math
import os
import random
import sys
import tempfile

ROOT = Path.cwd()
if not (ROOT / "src" / "research_lab").is_dir():
    ROOT = ROOT.parent
assert (ROOT / "src" / "research_lab").is_dir(), "Run from the repository or notebooks directory"
sys.path.insert(0, str(ROOT / "src"))
os.environ.setdefault("MPLCONFIGDIR", str(Path(tempfile.gettempdir()) / "research-lab-mpl"))
SEED = 2026
random.seed(SEED)
import matplotlib.pyplot as plt
from IPython import get_ipython
from IPython.display import display
get_ipython().run_line_magic("matplotlib", "inline")
```

```python
from research_lab.mathematics import numerical_integral, solve_linear_system, optimize_scalar, solve_ode
from research_lab.visualization import plot_function

integral, error = numerical_integral(math.sin, 0, math.pi)
solution = solve_linear_system([[2., 0.], [0., 4.]], [4., 8.])
optimum = optimize_scalar(lambda x: (x-2)**2, (0., 4.))
trajectory = solve_ode(lambda t, y: [-y[0]], (0., 2.), [1.], points=40)
assert abs(integral - 2) < 1e-9
assert all(abs(x-2) < 1e-10 for x in solution)
assert abs(optimum["x"] - 2) < 1e-5
assert abs(trajectory["y"][0][-1] - math.exp(-2)) < 1e-3
print({"integral": integral, "estimated_error": error, "linear_solution": solution, "optimum": optimum})
fig = plot_function(lambda t: math.exp(-t), 0, 2)
display(fig)
```

## Interpretation and limitations

The selected numerical methods reproduce reference solutions within the given tolerances. The quadrature error is an estimator, not a universal certified bound. Stiff ODEs, ill-conditioned matrices, or multimodal objectives require different methods, diagnostics, and tolerance choices.
