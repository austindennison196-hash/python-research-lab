# Recursive trees and combinatorics

**DEFINITION.** A rooted ordered tree has an ordered tuple of child trees. Its binary encoding is an opening `1`, child encodings in order, and a closing `0`. A tree with $n$ nodes corresponds to a primitive Dyck word with $n$ pairs.

The candidate `zero_split` adds a new last child to a selected node. Removing a suitable leaf reverses such a step, giving the reachability argument in the research documentation.

## Interpretation and limitations

The chain and branching tree are distinct ordered structures even though they contain the same number of nodes. Forgetting sibling order or adding a virtual root changes the mathematical object and sometimes its counting convention. This notebook compares a precisely defined model, not every possible meaning of structure.
