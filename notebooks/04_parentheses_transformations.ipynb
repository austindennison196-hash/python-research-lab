# Parentheses transformations and generation histories

The supplied histories are $x\to(x)\to((x)x)$ followed by `(((x)x)x)` or `((x)(x)x)`.

**DEFINITION.** In the chosen tree grammar, `x` terminates each node's ordered child list. It is not a free placeholder that can be replaced arbitrarily while retaining the binary grammar. We generate transformations through the binary engine, then convert representations.

```python
from research_lab.structural_logic import explore, binary_to_notation
from research_lab.visualization import plot_transitions

history = explore(3, rule="zero_split")
for n, stage in enumerate(history.stages):
    print(n, [binary_to_notation(word) for word in stage])
assert {binary_to_notation(w) for w in history.stages[3]} == {"(((x)x)x)", "((x)(x)x)"}
fig = plot_transitions(history.stages, history.transitions)
display(fig)
```

## Interpretation and limitations

The diagram shows explicit predecessor relationships, rather than merely a list of generated strings. Multiple rewrite positions can lead to a single deduplicated state, so histories and state counts must be kept separate. These transitions belong to the chosen candidate rule; other grammatical interpretations remain possible.
