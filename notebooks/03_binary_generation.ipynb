# Binary sequence generation under competing rules

The starting history is $0\to10\to1100$. Candidate `zero_split` treats the seed separately, then replaces one selected `0` with `100`. Literal `zero_occupancy` uses `0→10`. Standard `dyck_insertion` inserts `10` at any gap after the seed.

**COMPUTATIONAL QUESTION.** Which candidates reproduce the supplied transitions, and what extra states do they produce? Identical early outputs would not prove equivalence.

```python
from research_lab.structural_logic import generate_stages
from research_lab.visualization import plot_comparison

rules = ("zero_split", "zero_occupancy", "dyck_insertion")
all_stages = {rule: generate_stages(4, rule=rule) for rule in rules}
for rule, stages in all_stages.items():
    print(rule, stages)
assert set(all_stages["zero_split"][3]) == {"111000", "110100"}
assert all_stages["zero_occupancy"][2] != ("1100",)
assert "1010" in all_stages["dyck_insertion"][2]
records = [{"generation": n, **{rule: len(all_stages[rule][n]) for rule in rules}}
           for n in range(5)]
fig = plot_comparison(records, metrics=rules)
display(fig)
```

## Interpretation and limitations

The candidates yield different state spaces despite related semantics. In particular the literal occupancy rewrite fails the second balanced example, and full Dyck insertion includes an additional generation-2 state. These are finite computations supported by inspectable rule definitions. They do not resolve which rule the originator intends.
